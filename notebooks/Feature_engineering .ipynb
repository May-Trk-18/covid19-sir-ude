```python
import pandas as pd

df = pd.read_csv(r"C:\Sqldata\covid_population_merged_model.csv")

df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Day"] = (
    df.groupby("Location")["Date"]
      .transform(lambda x: (x - x.min()).dt.days)
)
```

```python
#Calculate the Removed compartment (R)
df["Removed"] = df["Recovered"] + df["Deaths"]
```

```python
#Calculate the Infected compartment (I)

df["Infected"] = (
    df["Confirmed"]
    - df["Recovered"]
    - df["Deaths"]
)
```

```python
#Calculate the Susceptible compartment (S)

df["Susceptible"] = (
    df["Population"]
    - df["Infected"]
    - df["Removed"]
)
```

```python
#Verify the calculations

check = (
    df["Susceptible"]
    + df["Infected"]
    + df["Removed"]
)

print((check == df["Population"]).all())
```

```python
int_columns = [
    "Population",
    "Confirmed",
    "Deaths",
    "Recovered",
    "Day",
    "Infected",
    "Removed",
    "Susceptible"
]

df[int_columns] = df[int_columns].astype("int64")
```

```python
model_df = df[
    [
        "Location",
        "Day",
        "Population",
        "Susceptible",
        "Infected",
        "Removed"
    ]
]

model_df.to_csv(
    "C:\Sqldata\covid_features.csv",
    index=False
)
```
