```python
import pandas as pd

pop = pd.read_csv(
    "C:\Sqldata\API_SP.POP.TOTL_DS2_en_csv_v2_33112.csv",
    skiprows=4
)

pop = pop[["Country Name", "2020"]]

pop.columns = ["Location", "Population"]

pop = pop.dropna()

pop.to_csv("C:\Sqldata\population_2020.csv", index=False)

print(pop.head())
```

```python


# COVID data
covid = pd.read_csv("C:\Sqldata\country_model_summary.csv")

# Population data
population = pd.read_csv("C:\Sqldata\population_2020.csv")

model = pd.read_csv("C:\Sqldata\covid19_modeling.csv")

covid_countries = set(covid["Location"].unique())

population_countries = set(population["Location"].unique())

missing = sorted(covid_countries-population_countries)

for country in missing:
    print(country)
```

```python

# Rename countries back to the Johns Hopkins naming convention
population["Location"] = population["Location"].replace({
    "Antigua and Barbuda": "Antigua And Barbuda",
    "Bahamas, The": "Bahamas",
    "Bosnia and Herzegovina": "Bosnia And Herzegovina",
    "Brunei Darussalam": "Brunei",
    "Congo, Rep.": "Congo (Brazzaville)",
    "Congo, Dem. Rep.": "Congo (Kinshasa)",
    "Cote d'Ivoire": "Cote D'Ivoire",
    "Egypt, Arab Rep.": "Egypt",
    "Gambia, The": "Gambia",
    "Iran, Islamic Rep.": "Iran",
    "Korea, Rep.": "Korea, South",
    "Kyrgyz Republic": "Kyrgyzstan",
    "Lao PDR": "Laos",
    "Micronesia, Fed. Sts.": "Micronesia",
    "Russian Federation": "Russia",
    "Slovak Republic": "Slovakia",
    "Somalia, Fed. Rep.": "Somalia",
    "Turkiye": "Turkey",
    "United States": "Us",
    "Venezuela, RB": "Venezuela",
    "Viet Nam": "Vietnam",
    "Yemen, Rep.": "Yemen",
    "Myanmar": "Burma"
})
```

```python
merged = population.merge(covid, on="Location", how="left")

merged_model = population.merge(model, on="Location", how="left")

print(merged_model.head())
```

```python
merged = merged.dropna(subset=["Rows"])
print(merged["Rows"].isna().sum())
```

```python
missing_model =merged_model[merged_model["Confirmed"].isna()]
print(missing_model)
```

```python
merged_model = merged_model.dropna(subset=["Confirmed"])
print(merged_model["Confirmed"].isna().sum())
```

```python
merged.to_csv(r"C:\Sqldata\covid_population_merged.csv", index=False)

print(f"Number of countries: {merged['Location'].nunique()}")
print(merged.head())
```
