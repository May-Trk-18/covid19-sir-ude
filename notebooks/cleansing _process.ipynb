```python
import pandas as pd

df = pd.read_csv("C:\Sqldata\covid19_merged.csv")

print(df.head())
print("Rows:", len(df))
print(df.dtypes)
```

```python
# Count duplicate and missing values 
duplicates = df.duplicated().sum()
Missing=df.isnull().sum()


print("Number of duplicate rows:", duplicates)
print("Number of Missing values rows:", Missing)
```

```python
df["Province/State"] = (
    df["Province/State"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

df["Country/Region"] = (
    df["Country/Region"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

print(df["Province/State"].unique())
print(df["Country/Region"].unique())
```

```python
missing_recovered = df[df["Recovered"].isnull()]

print(missing_recovered)
```

```python
condition = (
    (df["Recovered"].isna()) &
    (df["Confirmed"] == 0) &
    (df["Deaths"] == 0)
)

df.loc[condition, "Recovered"] = 0
print("Missing recovered with confirmed = 0 and Deaths =0:", condition.sum())
```

```python
condition1 = (
    (df["Recovered"].isna()) &
    (df["Confirmed"] > 0)
)

print("Missing recovered with confirmed > 0:", condition1.sum())
df = df.loc[~condition1].copy()
```

```python
condition2 = (
    (df["Recovered"].isna()) &
    (df["Deaths"] > 0)
)

print("Missing recovered with confirmed > 0:", condition2.sum())
df = df.loc[~condition2].copy()
```

```python

print("Missing recovered with confirmed > 0:", condition1.sum())
print("Missing recovered with Deaths > 0:", condition2.sum())
```

```python

print(df["Recovered"].isna().sum())
```

```python
df["Location"] = df.apply(
    lambda row: row["Country/Region"]
    if pd.isna(row["Province/State"])
    else f'{row["Country/Region"]} - {row["Province/State"]}',
    axis=1
)

print(df.head())
```

```python
df.drop(columns=["Country/Region", "Province/State"], inplace=True)
```

```python
print(df.head())
```

```python
Missing_new=df.isnull().sum()

print(Missing_new)
```

```python
df.drop(columns=["Lat", "Long"], inplace=True)
```

```python
df["Recovered"] = df["Recovered"].astype("int64")
df["Date"] = pd.to_datetime(df["Date"])
```
